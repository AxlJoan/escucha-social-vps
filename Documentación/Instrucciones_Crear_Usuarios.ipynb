## Para crear un usuario normal
1. Abre tu terminal y navega hasta el directorio de tu proyecto Django.

2. Ejecuta el siguiente comando para abrir la consola de Django:

```python
python manage.py shell
```

3. Una vez que estés en la consola de Django, necesitarás importar el modelo de usuario de Django. Ejecuta el siguiente comando:

```python
from django.contrib.auth.models import User
```

4. Para crear un nuevo usuario, utiliza el siguiente comando:

```python
nuevo_usuario = User.objects.create_user(username='nombre_usuario', password='tu_contraseña', email='correo@example.com')
```

- username: Cambia 'nombre_usuario' por el nombre de usuario que deseas asignar.
- password: Cambia 'tu_contraseña' por la contraseña que quieras establecer para el usuario.
- email: Cambia 'correo@example.com' por la dirección de correo electrónico que quieras asignar al usuario.

## **IMPORTANTE:**

Asegúrate de que el nombre de usuario coincida con el nombre que figura en la columna "cliente" en la base de datos ya que el código utiliza el username como parámetro de búsqueda en un usuario normal.
